```
import pandas as pd
```

```
df = pd.read_csv("2017_jun_final.csv")
```

```
display(df.head())
```

```
print(df.shape) # розмір
```

```
print(df.dtypes)
```

```
missing_fraction = df.isnull().sum() / len(df)
print(missing_fraction)
```

```
cols_with_missing = df.columns[df.isnull().any()].tolist()
```

```
cols_with_missing.remove('Мова програмування')
```

```
df = df.drop(columns=cols_with_missing)
```

```
missing_fraction_new = df.isnull().sum() / len(df)
print(missing_fraction_new)
```

```
df = df.dropna()
```

```
print(df.shape)
```

```
python_data = df[df['Мова програмування'] == 'Python']
```

```
print("Розмір таблиці python_data:", python_data.shape)
```

```
grouped_data = python_data.groupby('Посада').agg({'Зарплата на місяць': ['min', 'max']})
```

```
display(grouped_data)
```

```
def fill_avg_salary(row):
    return row.mean()
grouped_data['avg'] = grouped_data.apply(fill_avg_salary, axis=1)
display(grouped_data)
```

```
print(grouped_data['avg'].describe())
```

```
grouped_data.to_csv("python_salaries_2017.csv")
print("\nФайл успішно збережено!")
```
